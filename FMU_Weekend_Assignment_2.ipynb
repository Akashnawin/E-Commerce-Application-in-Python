```python
def assign_grade(marks):
    if marks >= 95:
        return 'A'
    elif marks >= 89:
        return 'B'
    elif marks >= 83:
        return 'C'
    elif marks >= 75:
        return 'D'
    else:
        return 'F'

student_scores = {}

file_path = "/content/testscores.dat.txt"

with open(file_path, 'r') as file:
    for line in file:
        name, marks = line.strip().split()
        student_scores[name] = int(marks)

print("Name\tScore\tGrade")

for name, marks in sorted(student_scores.items(), key=lambda x: x[1]):
    print(f"{name}\t{marks}\t{assign_grade(marks)}")

average_marks = sum(student_scores.values()) / len(student_scores)
print(f"\nAverage Score: {average_marks:.2f}")
```

```python
with open("alt.txt", "r") as infile:
    words = infile.read().split()

a_list = [w for w in words if w.lower().startswith('a')]
b_list = [w for w in words if w.lower().startswith('b')]

a_list.sort(reverse=True)
b_list.sort(reverse=True)

with open("a_sorted.txt", "w") as a_file:
    a_file.write("\n".join(a_list))

with open("b_sorted.txt", "w") as b_file:
    b_file.write("\n".join(b_list))

print("Files created: a_sorted.txt and b_sorted.txt")
```

```python
with open("fruit.txt", "r") as f:
    fruits = [line.strip() for line in f if line.strip()]

print("1. Sorted by name (case-sensitive):")
for fruit in sorted(fruits):
    print(fruit)

print("\n2. Sorted by name (case-insensitive):")
for fruit in sorted(fruits, key=str.lower):
    print(fruit)

print("\n3. Sorted by length of name, then by name:")
for fruit in sorted(fruits, key=lambda x: (len(x), x)):
    print(fruit)

print("\n4. Sorted by 2nd letter, then by 1st letter:")
for fruit in sorted(fruits, key=lambda x: (x[1] if len(x) > 1 else '', x[0])):
    print(fruit)
```

```python
import re

fname = input("Enter filename: ")
pat = r"[^\w']+"
counts = {}

try:
    with open(fname) as f:
        for line in f:
            for w in re.split(pat, line.strip().lower()):
                if w:
                    counts[w] = counts.get(w, 0) + 1
    print("\nWord Frequency List:\n")
    for w in sorted(counts):
        print(f"{w:<15} {counts[w]}")
except FileNotFoundError:
    print("File not found. Check the name and try again.")
```

```python
import os

# ----- Define the Knight class -----
class Knight:
    def __init__(self, name, data_file="knights.txt"):
        self._name = name
        self._title = None
        self._favorite_color = None
        self._quest = None
        self._comment = None

        # Check if the data file exists
        if not os.path.exists(data_file):
            raise FileNotFoundError(f"{data_file} not found.")

        # Read the knight data from file
        with open(data_file, "r") as f:
            for line in f:
                parts = [p.strip() for p in line.strip().split(",")]
                if len(parts) != 5:
                    continue  # skip invalid lines
                n, t, color, quest, comment = parts
                if n.lower() == name.lower():
                    self._title = t
                    self._favorite_color = color
                    self._quest = quest
                    self._comment = comment
                    break

        # If knight not found
        if self._title is None:
            raise ValueError(f"No knight found with name '{name}'")

    # Read-only properties
    @property
    def name(self):
        return self._name

    @property
    def title(self):
        return self._title

    @property
    def favorite_color(self):
        return self._favorite_color

    @property
    def quest(self):
        return self._quest

    @property
    def comment(self):
        return self._comment

if __name__ == "__main__":
    try:
        k = Knight("Akash")
        print("Name:", k.name)
        print("Title:", k.title)
        print("Favorite Color:", k.favorite_color)
        print("Quest:", k.quest)
        print("Comment:", k.comment)
    except Exception as e:
        print(e)
```

```python
import os

class Knight:
    def __init__(self, name, file="knights.txt"):
        self.name = name
        self.title = self.color = self.quest = self.comment = None
        if not os.path.exists(file):
            raise FileNotFoundError(f"{file} not found.")
        with open(file) as f:
            for line in f:
                p = [x.strip() for x in line.split(",")]
                if len(p) == 5 and p[0].lower() == name.lower():
                    _, self.title, self.color, self.quest, self.comment = p
                    break
        if not self.title:
            raise ValueError(f"No knight named '{name}'")

if __name__ == "__main__":
    names = input("Enter knight names (comma-separated): ").split(",")
    for n in map(str.strip, names):
        try:
            k = Knight(n)
            print(f"\n{k.title} {k.name}")
            print(f"Name: {k.title} {k.name}")
            print(f"Favorite Color: {k.color}")
            print(f"Quest: {k.quest}")
            print(f"Comment: {k.comment}")
        except Exception as e:
            print(f"Error for '{n}': {e}")
```
